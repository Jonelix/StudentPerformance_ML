```python
import pandas as pd
import os

def transform_data(input_file, output_file):
    # Read CSV
    df = pd.read_csv(input_file, delimiter=";")
    
    # Mapping categorical values to binary
    df['school'] = df['school'].map({'GP': 0, 'MS': 1})
    df['sex'] = df['sex'].map({'M': 0, 'F': 1})
    df['address'] = df['address'].map({'U': 0, 'R': 1})
    df['famsize'] = df['famsize'].map({'LE3': 0, 'GT3': 1})
    df['Pstatus'] = df['Pstatus'].map({'T': 0, 'A': 1})
    df['schoolsup'] = df['schoolsup'].map({'yes': 1, 'no': 0})
    df['famsup'] = df['famsup'].map({'yes': 1, 'no': 0})
    df['paid'] = df['paid'].map({'yes': 1, 'no': 0})
    df['activities'] = df['activities'].map({'yes': 1, 'no': 0})
    df['nursery'] = df['nursery'].map({'yes': 1, 'no': 0})
    df['higher'] = df['higher'].map({'yes': 1, 'no': 0})
    df['internet'] = df['internet'].map({'yes': 1, 'no': 0})
    df['romantic'] = df['romantic'].map({'yes': 1, 'no': 0})
    
    # Feature expansion for Mjob
    for job in ['Teacher', 'Health', 'Services', 'At_home', 'Other']:
        df[f'Mjob_{job}'] = (df['Mjob'] == job).astype(int)
        
    # Feature expansion for Fjob
    for job in ['Teacher', 'Health', 'Services', 'At_home', 'Other']:
        df[f'Fjob_{job}'] = (df['Fjob'] == job).astype(int)
        
    # Feature expansion for reason
    for reason in ['Home', 'Reputation', 'Course', 'Other']:
        df[f'reason_{reason}'] = (df['reason'] == reason).astype(int)
        
    # Feature expansion for guardian
    for guardian in ['Mother', 'Father', 'Other']:
        df[f'guardian_{guardian}'] = (df['guardian'] == guardian).astype(int)
        
    # Drop original categorical columns that were expanded
    df.drop(columns=['Mjob', 'Fjob', 'reason', 'guardian'], inplace=True)
    
    # Consider dropping 'age' column (uncomment if needed)
    # df.drop(columns=['age'], inplace=True)
    
    # Save the transformed dataset
    df.to_csv(output_file, index=False)
    print(f"Transformed data saved to {output_file}")

# Example usage
input_file = os.path.join(os.getcwd(), "data/student_portugal/student-mat.csv")
output_file = os.path.join(os.getcwd(), "data/student_portugal/transformed_student_data.csv")
transform_data(input_file, output_file)
```
