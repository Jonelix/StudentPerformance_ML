```python
from sas7bdat import SAS7BDAT
import pandas as pd

# Define the file path
script_dir = os.getcwd()
relative_path = "data\\pisa_2022\\CY08MSP_STU_COG.SAS7BDAT"
file_path = os.path.join(script_dir, relative_path)

swe_rows = []

with SAS7BDAT(file_path) as reader:
    # Decode column names from bytes to strings
    colnames = [col.decode('utf-8') if isinstance(col, bytes) else col for col in reader.column_names]
    
    # Find the index for 'CNT'
    cnt_index = colnames.index('CNT')

    for row in reader:
        cnt_value = row[cnt_index]
        if cnt_value and str(cnt_value).strip() == 'SWE':
            swe_rows.append(row)

# Create DataFrame
swe_df = pd.DataFrame(swe_rows, columns=colnames)
swe_df.info()
```

```python
import pandas as pd
import os

# Define file path
script_dir = os.getcwd()
relative_path = "data\\pisa_2022\\CODEBOOK.xlsx"
file_path_xlsx = os.path.join(script_dir, relative_path)

# Define the sheet names
sheet_names = ['CY08MSP_STU_COG', 'CY08MSP_STU_QQQ']

# Read the sheets and extract relevant columns
dfs = []
for sheet in sheet_names:
    df = pd.read_excel(file_path_xlsx, sheet_name=sheet, usecols=[0, 1], header=None)
    df.columns = ['NAME', 'VARLABEL']
    df.dropna(subset=['NAME'], inplace=True)
    dfs.append(df)

# Combine both sheets into one DataFrame
combined_codebook = pd.concat(dfs, ignore_index=True)

# Optional: drop duplicates if needed
# combined_codebook.drop_duplicates(subset='NAME', inplace=True)

# Save to CSV
output_path = os.path.join(script_dir, 'data\\pisa_2022\\combined_codebook.csv')
combined_codebook.to_csv(output_path, index=False)

print(f"Combined codebook saved to: {output_path}")
```

```python
import re
# Define patterns to filter out
exclude_patterns = [
    "(Raw Response)",
    "(Total Timing)",
    "(Time to First Action)",
    "(Number of Actions)",
    "(Number of Visits)",
    "(Number of Short Visits)"
]

# Filter out rows where VARLABEL contains any of the patterns
mask = ~combined_codebook['VARLABEL'].str.contains('|'.join(map(re.escape, exclude_patterns)))
filtered_codebook = combined_codebook[mask].reset_index(drop=True)

# Save the filtered version to a new CSV (optional)
filtered_codebook.to_csv("filtered_codebook.csv", index=False)

# Preview result
print(filtered_codebook.head())
```
